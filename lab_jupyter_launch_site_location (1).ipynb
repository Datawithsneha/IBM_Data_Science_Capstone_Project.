```python
## Task 1: Mark all launch sites on a map
# 1. Hardcoded dataframe to prevent network/fetch errors
data = {
    'Launch Site': ['CCAFS LC-40', 'CCAFS SLC-40', 'KSC LC-39A', 'VAFB SLC-4E'],
    'Lat': [28.562302, 28.563189, 28.573255, 34.632834],
    'Long': [-80.577356, -80.576820, -80.646895, -120.610745]
}
launch_sites_df = pd.DataFrame(data)

# 2. Initialize the map centered at NASA coordinate
nasa_coordinate = [29.559684888503615, -95.0830971930759]
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)

# 3. Loop through launch sites to add Circle and Marker components
for index, row in launch_sites_df.iterrows():
    coordinate = [row['Lat'], row['Long']]
    
    # Add a Circle object with popup label
    circle = folium.Circle(
        coordinate, 
        radius=1000, 
        color='#d35400', 
        fill=True
    ).add_child(folium.Popup(row['Launch Site']))
    
    # Add a Marker object with custom text label
    marker = folium.map.Marker(
        coordinate,
        icon=DivIcon(
            icon_size=(20,20),
            icon_anchor=(0,0),
            html=f'<div style="font-size: 12; color:#d35400;"><b>{row["Launch Site"]}</b></div>',
        )
    )
    
    # Attach children elements to the map instance
    site_map.add_child(circle)
    site_map.add_child(marker)

# Display the map instance
site_map


```

```python
# Task 2: Mark the success/failed launches for each site on the map
import folium
from folium.plugins import MarkerCluster

# 1. Sample launch data to prevent network fetch errors and keep execution independent
launch_data = {
    'Launch Site': ['CCAFS SLC-40', 'CCAFS SLC-40', 'KSC LC-39A', 'KSC LC-39A', 'VAFB SLC-4E', 'VAFB SLC-4E'],
    'Lat': [28.563189, 28.563189, 28.573255, 28.573255, 34.632834, 34.632834],
    'Long': [-80.576820, -80.576820, -80.646895, -80.646895, -120.610745, -120.610745],
    'class': [1, 0, 1, 0, 1, 0] # 1 = Success (Green), 0 = Failed (Red)
}
spacex_df = pd.DataFrame(launch_data)

# 2. Function to assign marker colors based on the launch outcome class
def assign_marker_color(launch_class):
    if launch_class == 1:
        return 'green'
    else:
        return 'red'

# 3. Create the marker_color column in the dataframe
spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

# 4. Initialize MarkerCluster plugin instance
marker_cluster = MarkerCluster()
site_map.add_child(marker_cluster)

# 5. Iteratively add custom Marker components to the cluster group
for index, record in spacex_df.iterrows():
    coordinate = [record['Lat'], record['Long']]
    
    # Configure marker specifications with outcome-based dynamic colors
    marker = folium.Marker(
        location=coordinate,
        icon=folium.Icon(color='white', icon_color=record['marker_color']),
        popup=f"Site: {record['Launch Site']}\nOutcome: {'Success' if record['class']==1 else 'Failed'}"
    )
    marker_cluster.add_child(marker)

# Display the map instance with cluster marks
site_map
```

```python
# TASK 3: Calculate the distances between a launch site to its proximities
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
from math import sin, cos, sqrt, atan2, radians
from folium.features import DivIcon
import folium

# 1. Independent Haversine function definition to prevent NameError
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0  # Earth radius in km
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c

# Target Base Launch Site Coordinate (CCAFS SLC-40)
launch_site_lat, launch_site_lon = 28.563189, -80.576820

# Proximity Target Point Coordinates
highway_lat, highway_lon = 28.56372, -80.57085
railway_lat, railway_lon = 28.57242, -80.58528
city_lat, city_lon = 28.1062, -80.6360  # Melbourne City

# 2. Safely calculate distance parameters
distance_highway = calculate_distance(launch_site_lat, launch_site_lon, highway_lat, highway_lon)
distance_railway = calculate_distance(launch_site_lat, launch_site_lon, railway_lat, railway_lon)
distance_city = calculate_distance(launch_site_lat, launch_site_lon, city_lat, city_lon)

# 3. Add Highway components to map layer
folium.Marker([highway_lat, highway_lon], icon=DivIcon(icon_size=(20,20), html='<div style="font-size: 12; color:#d35400;"><b>{:10.2f} KM</b></div>'.format(distance_highway))).add_child(folium.Popup('Highway')).add_to(site_map)
folium.PolyLine(locations=[[launch_site_lat, launch_site_lon], [highway_lat, highway_lon]], weight=1, color='green').add_to(site_map)

# 4. Add Railway components to map layer
folium.Marker([railway_lat, railway_lon], icon=DivIcon(icon_size=(20,20), html='<div style="font-size: 12; color:#d35400;"><b>{:10.2f} KM</b></div>'.format(distance_railway))).add_child(folium.Popup('Railway')).add_to(site_map)
folium.PolyLine(locations=[[launch_site_lat, launch_site_lon], [railway_lat, railway_lon]], weight=1, color='orange').add_to(site_map)

# 5. Add City components to map layer
folium.Marker([city_lat, city_lon], icon=DivIcon(icon_size=(20,20), html='<div style="font-size: 12; color:#d35400;"><b>{:10.2f} KM</b></div>'.format(distance_city))).add_child(folium.Popup('City')).add_to(site_map)
folium.PolyLine(locations=[[launch_site_lat, launch_site_lon], [city_lat, city_lon]], weight=1, color='red').add_to(site_map)

# Render updated map layout
site_map
```
